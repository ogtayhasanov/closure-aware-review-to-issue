```python
# COMPLETE KAGGLE TWO-GPU TRAINING: seeds 42 and 67
# Paste this entire file into ONE Kaggle code cell.
# Launch only with Save Version -> Save & Run All.

from pathlib import Path
import base64, os, re, subprocess, sys, time, zlib

SEEDS = (42, 67)
SCRIPT = Path("/kaggle/working/splade_bottom15_resumable.py")
PAYLOAD = """c-rNCYj@kovEX<83Iy5g01gR@8heuA9p+wVteN$R?fBZBxx1_3!GTCnLIeT~0NPd*?Qh?D^rN}~P_#2&_N<SOO#t21)z#J2_3G~5{^s$%Z6041)#JSS5bSQd>$>{-Bn-p%O;%JvR;`2lLsssyE)U+ldHL-5`Sjg+P**{Bod;Rj<k|Xmx`O|@d>vfYT~}|Xv!}nCZi=$Y8+f<Qx-2<4d4FBB!H29VvZ@QR^|}r6UA?-VUe^0+oi(>XlehbA9<=-2u5P+uQ#ZkS-(;6%9<0iGzYdyx)h0pkqC07`4|z)nZ0mJiPJ(*Z72Dz;c{2&x)pfq!m-zGT>mP%*1H{@%P}bQxZ-T45%A2gK;Va++|KCndu#1;@+fAY7>Sk9LfZ#UoGDJMcHqZ;8Kn@rJ847q=K<5c}=+$4ct1Cco-)2{N5Y<&46xBN4<?s(csp>Ajtm~V{ZQiz6Z3nR0JYUE2lam1c{D$crJiZ|~k3ZJUO;KGvZg(Z1n=(0PPt)zb><X-y>~4eU6n_P?ui@9axt#)Ceg$B^4n91HCnrA?RaR10+c|uh1ZT5J@Z@X~{P7QyV0L;6<M<wik4Ox_i!K1}0Kx;+fKFLi-UgfE6F|*iwCOvw@#O5^fc4;Ghs+6>1r3Z|y{&-0!Rk7zu7EEt^X_AwSIjTSc)(Pvrr32s{%M!5I_8`<17=ghS5&(_yxzCS!nXnU)-(}4iIZY?TV2k9MvnqRgtFW|)`ijz8YuIwO;+uPezEqpf;PKimF}{lX;VEpOlibuQ}3IgZi*|SUD3AtJox_I&#fuYT{n3Soq(!X;!(QIy45wbj$rgGMK(jSwt&pZ$qQgYWZqx}G;Luv2QOc~qDjrKo1CXE(}dBYITpazVpGtEti!J-L?u1tgH>I1(Dyd@!{pm1PXG`;0sa2Ihl+ViZG3Y!c{+O<XxdD$xw>k*W)Guq0vM@YcF?Cbh>B`e?$=22>#SK%tNaSO^C9O#Z9KsN0>IS7?Q~tVf1|ldjB+9-Co#-~Fkndmqud~{JlK`{7H8p<U`U$P`wp2PLD6?|vT5pVkfxh`w{P+^4T>!ZrL2M}P33{M`dxFi%bGUV&(~RdT^5)6@o#Ni>EAG3*ZOze>R-&l=E(qRulHS1nx|V+0c2gy814Wdw58E|gYfA?ce_J6s`t-}70}}aj4g`ON$?Ux-z0dI?Xb`!c$Wh;s#T7^;QjkOjF#zMwco-V&)T5c=~p{I7+%2N-P+((H>+#sA*m{=R{3v}&AwXkG{*uzD2(YV_P9+*ENI-}_e&Dq6Z#E&tttTBT398vdDH5`5lOLU`>uY43&Rg}^D1j^^7YGC6Z7SLeUn$TR?w@fyh}^q+^k9U(vf~Fy6f~~)@=8?$w_>2a&rDZ-<*H{{``6R_Wb>`7q8*RfBtm-_Qm<TU=jQw1^$;MUj6lFr@$<Ia{AO3c>ngro0sSB@yj=7Da?<PSATte{xW_2?A1BE3oBsRMoHZ1s)h*zYe@?%i~nw6{DAR+O2DKi>wFUs@1?+mZ4?L7KN)sRUW0&ZciAeRQ$Oit0}Yr0&zdWc)m8U~zC^JomcRmyFzI^~vV;#O$X59R<{ZxCzwZl>V2k(79vJ63FLw)88IbR&_+fxJNBdx*M;WXwizjCXYm+FNHg(<U+Q|C0IzC0AUd5<%7$1bl)Bk9YYGHf;C}|1pU<bcGSn(1C`?Liv9jttqI9E-!o@T4nemhus_RYcCFyE)xJD62p(6-V?pk_u>d=a9OmBaG5$F4A&{LG?SE&sTlKfo(D#cqnCdKm0@UHX83LamvE6Z38eLGzIjeak8-iyK%U%dG8S;oIVB21^|cZ}9ONbdAjFs#b~2Kfz3G6;#F&mdSlnX=LTxRnr-li*yY#H>xlHHV@#R9A!5O`Z>vWn&huR)6~<tyGA(B-DG>SE}DoRS~W2VhqS(-N5%oA8`rBrQH{cnKvod{HLkpiaNljFe+*-k%%D%LOVtkkLEI9J+eCeykj7g9V<40km&>eO6~*F*tZZ}Lt@b)Qd-_deyD<Sm?W>y<d3=tV9}3FgyYGUtQ|cio3Ec3y0G$rMs&zF{poaU>Tn?j-)&w;OO$7xkgi)DoFW1>zl}u2ni7ZN>Y+Qz6YyntQlKl>umkT?is=i<6pZJ#<)CyLqv{*-|ZRC1T-EI~j$=6BIRvDKPO)Kc2mZ*Y73wl?V!HUggC#X6BlT^C|Rp_g{iTl-|B`l?&#|L0-z*b^t^b5Emj!BuKM+E>-O;uyjagCvp2g8V;6o?iT!^oi0<e(XVwv4hbYQpKi|FZu5FUcuNA*kAK1m9lFr)SF(@1>&ps?4(r7WJ&mnnR>F^z9EqkOL17q0B9y_K|fd>8DXwbY*U5>}9>awNGVs3804Cr;4c}kApOQVfrAr!7})Pi*Or*h4_=htef!8Hhlj}^-FbU+Mvgm3gtTi_dfI~0F&REg(!Y#u}yIu74iyyaA#M*NF@9LGOD2$_-P6FUhpvDL{;`eiu0Q0s{xrbS7m(}g@b07oUSYc0jz|j<fJJC!>y!c@Hhw$*=S*$urQB2kzl)%fHn<gzXBMaXs!l>L3Y|QaRvr-@Iz7Nuj}pybb_2WP2EHe`0#ZNQU$Cg^i%{;1~fQFtFdMn48nCVNS7dVVUk^RsI#WH>cww^H%-2Qrpe{gbj1eNPBf;=Vg)R88&sfbt{sb)<VA-oBi7M=kgr+=4=fB+>DR$mfanK6^eb(%g7Nu15ChCXprbZI*P!1+3k6URS|gVXXy@z1<EW$s8^99qCcj;<ln-|Ez$NT1L7ao%fW9r!78(GNN?{k$C>AW&c5zbHA5rYNg-cj}_`CQJi?x5(phyh=tbJsXEtC>N%aneL4GwXRf)jvXG@qRoeriz%H@-N_L33H9y*OPu62bW*=Xd@I#v55A@6B$Ib1-U1%Q+Iqo9rJmSRvLak$PcDYVye<ZI^GM2r1dR97@}K>y)Oo0StOaT#So}N4{7vUDP;{Sm+_eOsbqi+YEh^RDQijmdBxC25!M>CFXB~b`Ppx){5%kWdoKRFAi-G&0<0Z`1{#vmG3(CtF-&grual^3|Ju0PMLke!9#zG1qF5=kf~UqgC=JpBm;d?M1*y;-Cx2A`b+!hmk1Zp_`kuxhaszx4pn0*41MY>oB@=qshV2I;d2F}8YwFkVI`0vQGI_{9&W-{KtiO+oo0-CB?<5Bh~I|}gM79m(DXowAUf@d1J^P5VRrWG2nIF(6z+ENwRs8iHpOK*T5sm4o1h2ehXyAH&NO)Oj=XI~hYWey$0{M>wcaGFy4-IocF2VTA9{M#dUGMcEEBMBnmp>ghOUF61e!8_;dk-Fcs$sg0dZ_)QB9jynl8~r7%qcHGz9BS%(lv>1puTUT8qKCoNCW2&)9UE?S^$3bf#UvcusK8myA(c@0%6Lr8{eLw#5~I3+F-jj($y_RqLiM)+sPF@KBk)MMo-01<Mz(I^Psj3@bi^hvTZ<e2e98rr#beMluOYO5r35-7CFClVI+37khYbBri;Ey8zHMU^*mVVQn2<iDdNRz;a^YSdp}8l#pJNd6oj2LXhB}YQMeAn-sVMJ!gzHzawSde^W6?GbKp757lQW2?bm=!JUe(sAI|Xm;6=>tk*e+bC6H6b+iE$SakHmTm+zZ*c%jG1|F&8$Vb%PMM%%U!q^%9J<^LomZx$;#=}f02)dLkTuAm_k&X5K@PXcQDT}dFcWH{|B?A63kORQle5A?&I_TNYEo%WFJp?>H6CMRN;gsoy82<z89>&PB$5It;CyKb*)bs2Frb@BKPK9?cZ|<je+AF}&;`?x*228;VT^A=`-{N%86#Zdca1*{{Mf=XuVLrK6Ar<O0VKZl5q_xm6q>`ov6D2KGq$N)?wEgx)xlWDUK`rnC+fx;l9cyhx{hq`sMIA2W5;qClCl#+?VZ?^?QpsP(Vq7?lplLU3E-L>Snm;4;9=4L2d|Q72Wnlz>{tO}yora>_6&X2W^|+yw3nl1x3r}Q6TTd#={>QdoJ=v-IY6h5{_?j-M7Aej;MxnB_$v+hNM_@y_ob2khzz|)kc~pMX@jL(Hz1ucoHQexElPZD_75TILHiQMgW7*kzO{#;$N@=HKgh{a4C77v^<s|+R`<kp+58iXxz-VG%+1~j8lCCZb^ciYZ)<;Fs$>Kp{dJGajDBhi74Hp#$hJ%MSN^25<V|20Y>TX@2aJ2pxeoj!SQescA>4g~SB^hP@%h8Hv#6>;(&aI|p8pWwo9j6{_uG&`x6p+42uNv41qd`G`6rQzh-e9m{z{9f%&nYPZMf5m8kk}@`&UnH?a|PTHdcf{{M4@UDyv;XzGNHg+C0VHRcRn^Xywxt9B=pP^=r}GLdJJQy*E){*FZb2}f?nEG0qotK(>is<^JRT$&wRlQ>(oc&1q`g%F}JdP!2S2zfKr-_KMa2d?H02d{=SDLvg7219go+9CUcd8cHCx7fhjDGiLawdpqh0-Z5r33t@35iKh8@BHiy|14KD6QB_Q?J4l?}^y|S$_QeGCE=ko33>nGm^Q<cH72xgNfr%w}MVhz&>2SqL`l1{Ee0lRz}d^ZB=J7)?xQS%`M`w(9-Z@s8UHPOvYb#r<Q0uDCa4WfhG<uaF&7$kR<aA2ZpnUP#I^?rAG8wuGFU-VZfD^I|X+f|tkjp-0f#bVrT1Z3BzD5fbldti!%&jN!$03wgvQk0?vy$cLFQ$+>Jwy`=MRfJm5t2NkqmawSr_GRNl%gcdoqPUfDhgf?#tQ#2+|B%xRB{MM48pM<{fmpxAm=kAX6|Mt$LO$ga!|3Hurcm(Wd23^~Ls?a5d+I4wmum%Yl>@wbTXZO%I-J`rrYo(%@XdFeO$7BSba<C_+rjjiJvlWJE0Hl6#~?LE&yZ#1ti5c?Ss8Sk6%O2MAz^H@Lshzs*xF%Uvm8WirxeZtFjuf#bX^kxjEO)#Ced&E!iKUik&SVpjTius=!G}#St{;yHbvQh8Tm+NP2b34)F*z7qRMlSghth5BiE{F7Xzy5G7kRf73!<3%B8hsA7px&L%)!4z4oRA$`D36<UukSzRW;t!<;o_Pm?06apN(Z^Sb9YiL0NUvwcq$O63QAST@99tWjVax0$&V-?CUA^85&MwO`AAR?4OF@d$NzcdPU+W&X4Mp>#awoe)dC3I|J*q_U#VX-*G@1Tern1i~9`Ra$eAgfPHaJ_Lu+wpn{)YU?gdti^@DZ97X2TE|IN-9{OxDvOy}UL-{NE`-5lw$_a#Wxcv^UHGD+I}!U7oebK7LpI>jf?6imn00Y>8h@4wd53e#=~cVm!cr;q#{g60N{|?<8(Qy@;9M>Ca}0E6>-%1mQz^UbBl!|d*VuU?%M~+AIWKula+=n5*T6%u(c|>kT6o6FbC%&aepo3m8O)sQ3JaQYX5zYN><JquB+C%jMq~L&RdPP)fU!#v29T^%V?bf1Tpd|NrR*G=e$>MV+NP8w6nznw);b#pLVwobW7|iquIq@ame4^jgP9b?o-s90jTLoM%dz8Jt@c3mUEZu<VZiFM)6?M5sA?QM4x;gN)ZdTRi|5JN=Kgm*1okbZt?7w_2)M4o$9*$Q^M~0b;ka$2Gb^O9a3o)eAXuo~2%OV}g*my)@n=bY6M3-w(!#Gbk=`SB@Ip#6i~QGa6*dfPjWKEtR5I3tvNp)701$)LMxO>6KjF6*g&e+zlFqa1R^#j*<v>NsOw2eb2kthz$y3UjPg&wnGHzs-FgMi5;HgZC3A=<Rza)DLoxUXL=A0o0+;E!APLKBM$|vQ|)q)VGc(qHirpaz2jOpg<zN}2Nvj#(vl(4F@te7Am{|E@D*(Y!5zjwGTD6)OxvhDSLvng|26AcDZApuy4Dii6lu^<Fn=8FJLx7mc*4P!ktsHOr;ZkKLxEP+_%RoA$f3-C+hpi`63z#=`0sne$s=J`dYfp{sqMeBqI;s5+AAjDIGCLfpqXp4~fD-uy%%*7Lh?^)4r?dmU(`3!Y~EzmJE^vcC%fWg{Xii7`7p9Ox-=ZDte|D?qitl}HoBc?Yeb!K;++_;ik!<>0@n^nbz^IzpOu|&CZ{b2^Lx#i$~5<}6!_lwZBOvdToQ8lnU$TlD08IsjN-wjlu$K5L$%TcqKxJ&X6w5!Y8!9ysyE3%Etg^Ii8=rgw?^p$SlE8(ui^-6na9}8b0D2=>^nfi{l81}NXoaTup(7fb;FKJ%XA8lTzv06+ga%N0CzA%|0xWM2h#SZs_MMg|}+2TqD;^)YUJu4i&54<M_+S-E&=Y0c);m@o6I>T(>>_b+Rs7pwlPZSYg(yT0d&0K=6o>wSv_v@-k);TP2ThL!yOnh1dIA6z0UxEgCz0Df5naJY-gjtnY+XhN0S6hc7^}j+9X_taj0E(u7P`Q~5Er6mwSSDe0df}OyN!H@xgQ-5Tk<o5NW9KjW#^w>#jhk!Hw={>Vi%tRqsy-4=0B`h?X=s%)9dv46<sIg97%Gt<wH(Hay!KDY%ov#zZa`m9_2tB+iu+b~O}#ba`Hph1&e34Vn+S&fRgHVL#s`$P2vas!S~n^0Kmh!Le{xEX7~B00leZF6-y=T(0@@F4FH6v&gPI>rT$m{RRvpjyb=H#K%>sti3pef*U<9Og4+)yMd4;pe-;7{uCI{<953~!bc(p06p!t29r>hKL9B7IQC#I;KZkqbAQebYbB0g<o@+?Q5!0+=4){7->Ec#HaV8-!lej=ul<t##ME`h;uli$t*T5W*eBd&*0mMXn0>lN<K;x(SuE+{<V!nlYLRys-P-nvHXtD;lsF{;DSqes*rruN}SyL+fpdzDe$!x9&vc|}eM(o0)l_xvAuG@aQ|-~|%3B<df)o$Vt2-Q?vy;)Y^W6Ai@P)(0h+Kk>c<Q1RErb}@?+epIUhZ{uQ#i!2?f7scS(AW9qLuvVdEe;oWSh|V58`uc2e*%0;P-eTi?WbAl}V=7a5+y^xu*uOaOHOwLNM68iRpPPLpGfwm?Z5R#vHk1Wp^O^l+7BNn2hnWWr@%S+carFxiu9r&JTJ84MHN>?O6zY_wPEdSZ^DP(z#Wq)(CU3g8TjM!ck73F0e|r8b`0>q8#&FL*;ev2Q$tSb3KdKEn+Z{%9_T=E6C8t^s#jXD;eHPy5(<dD2$g_r$YA^@{AI|dWQ)Z;RxVrArb-v1O$re1FsZ?*=P>aQjTG@z<jTKusE!0q<0wq4Mj)ccF(8q;S35w#EP=#Q|%HS8Bwu2k{boPx`Ons!2Ba)q}A<7)H&rY^NfV|i;agI9i0*wHH(17QxYi_-Ws0&$3l5E`%iwek--p6gWOF>v>yb}`nuV03TvrTcICP$MdUYOLR<$a>W_V>WD#F5A1EKwXlsTWd?%n&1w@QOq^Qe(Sxj@oO(in`kJ$yhS9;Bw#D?Vqd#6s>lT(*`L<=RO=Ej0*Bf3S6*dMG_Hh%-<{mBTa^o{}~0ln&%2fvD37Z8?S?Powqj0UY*YoK7fw(g*fO~ug_;ACC=^WV`R5~J9xP`sXald=O$btdQyB}_>RQo=VSXZYXg0Z4MpeH#b%S!<}hQY9f7Ljp&UTe{r;<>NuULetb%e%wp4jBSg1JG%NF%=YtG;O^y<ayAGsZW+w(x?k)X#{ROqWEx{oP*?tRAnIjC2_Cn)4+=eYU3vLArEJuN|wh7hPOQI+4}b!~gV`|>4KDI@#eaMm{+Z@;1Fzov1!YDi<$%V(#LkWkiuIZ|T83}D=aLA9#O5*?6AJ=O<sTnyP!(!)g7)5OG!LEW_`A*O3QB7h<bm=rCAT||i=qBVr>OG>R@st)x<Bw3aODekWoyKKGIig*8O2U9@__Td6nEm_!~lYu1-kxV<X4BABqYxHC&b7!DK5u3IX`$E%FzSC4yaLl1mX4(2tzfsneDd9LvJoV6%89E+^Y5{M~;WMgI1B}asoOo#~+7YwOqUfQIo>U(Jsj?c>*PD-ssTagSMxw3o;hZOzeIw!$ijmVz1;x~ZOab0+Q+=3(0wZ3@zz&)M`L=Fup`2xodq-aqRf9hTry{3C_TwVdW)52MfEdOoeN$+xDk#7$FgP018t<icJ_$2KohG&rr_<P(;(45;)6w==`S!huV8&@-4=OVsA3buAmB&i3_47GcNnd0iD^JZEY5BxRXK2&)%Q$k++UJa|IjzmiCwh}STW8z<id?ymd$s|`4n{)mt1Fv@V}IB$()w7)2T3o30j)r73d5fAqOhPtnvzvsl#z8AbN6Fd%!m=u;gZq`E82nCsl^Ou(Kzr&T%C`Fkv5U8ic6IQ_Q)}sRFFy^aL4IehYl~Yw6q80$-PslY<?jh++t?`1Q!$-N}?W+Z}UDPr22Eol=2Q;!DY&6S><&7fy$;FR#w|B&`NMb+@nWQGZV51f$ULUg-`&y9XiokqqWbi#x84HJ$mlz(4v_24?;1k(ufl3@SwTmnaO7hXyY3scHNg8?wR|1EZYTraz#P3LrrL29t2?*Z>Tu5l83_A07b-;OG)c_z-YgG7Udj40OdvmLnws-6zj_dE+r-$mJ*LWHgGcwM&gGyJR}49EPq`9BN4`70NmO@@_<}#`YjGrkV}EAFe=BO?&vp(A&CjEB#yQ(K2&K~2+|L@{t|=o!jS4WY3laf9Q`J$q$wW*m!^?ZB#cMk@M)I<LVMdL>PZyiwM(V4sD(^UvEAO_OU^x%%3O>lcyt;|ZN{GPq&@G#<ApyRkA}4w%DQb+eRN(lY6-|qw>m`Lt~hCxal~vn4q-hK_1S@J#lDWDkJr^jdUcH-UDfHb)lGEaVl(ktSZXLh1tAVjdK)|$Ch$|;Dpz@e@(78nvOlzE36D19VOU7*gO&pdTqLUoX$sP@!Au5*W;Srzre{!?d!20atcv2`_l9uxjqZ<ngB|%=`B3E{8#xqpeX5V3g$aDXkY~2K#vS$L%semy{D(R`u})FoMzP@vy%{qSaX58=ll*V#nbQd#rxwW}143}{K%L9O)$B<dOty#1l(UD+P&NU{H!`_e%g5%0;SPRbQxjMfbFEe#kmX>43)vrfIWbI?Uh_k>gssiBJRFB>A!TuNC(K93&f*>nMOA&E2A{=+A1IC$#j#t(43%2}xNNYv88r|7tY4h@hgKQEK#IM+TwMZE?RZLvUc{d7z%6KSKU~b<?4GM(rUst<s^UxX+&NmNIEO6|ZYcEzAf||@%4PFnWah-ZBahm%^x}UBCE*@omv@_OKnj8;01W;Q>mP|l7KQdN4FQOq1NwMramaSp^Pq%!9rxC3yQW#fKq>JwP}Zgq5#_?UA>u&chxF&q)T8YhJ?tu;Kj0@m0LjZWJ><r9q1?NV6l9W{&^fR~%lkOrN`uCZrr=%cGw^xXX7^*X%o`oW*IuC^xOtiF+I*dEi)v5DwRsiQ`RPi`4{pv^*5S|`x<xte*khdM%DaZFMmF=Z*A6i5u}m8d(FH0Dy&kHmQLqm_jxOa8TlY=xl(C2B#AnQhUuQ1->+?X^@WNn7gw-DVeKdCaT-f>tDvY={=fX33Wjv#PWu^T$dP-vFIAsCq3|FDUu$>w<31Yk<Hf>}r*5<H=dE2XnriA_6_Du})jne7->S}sIHFME{o3E)u{&kaqx}-wkG7e3SBtD=l3X@%zblY7Rd#QaWtRgj?CmuXow#L{DD7`BTLdYT$*`iIy_5Db$^f*HDxccVM%*G@-QYXTq+K6u1S&cagLsHgxPaz{ye%6zoysb`&bIyhs2yv+%p?-+r6(<7$hB`8%Ie~7)@o^n<ePV+`Z#mNMet1|&K2gb~O23wr$kLDJ5UpQLw*#iq!xZgzmPzJI;R(_{mRrqKdQvwOIOz!ofxb+98V<Yoa$?z=HqO-d5(7MO(W^`npzTQj7m6=Fv?r8k&naHTHc#*$v>ivt`zZJZIaX$H|5BM_-q};JZ`i3Q^QX;G(XH`v*khvgQR0}Z8lUuS3X_yC*yK;r4ca>k^I2I^x)U%d?%me-ap~Y=-sIxm3Ka!T#H7a%2H`M7X%JZ5EvGy74IZ75WRFT=d@*K++x{ZaiHYn3@^(Dr7f06yTlutx*J%a9&oc<5GuP}ou|f$CW^Sw{GiryMa+o0SaH(>3pp@QG_Dgk61h}Ja-aoz*eRhsILt3#vnKFDQLG*bpff#9=!%U-72waOtMs3`6k%nU-w95GZFINIt;RA6jN~ovAUAO!Ir~Mw+Id}87<hNrz6oUDV^HB&s9QRUqo6=LtbRjmQ6KD;);o+sd_?}yNGdRr_KIsD5P%j#q_iW1F2H)2Wp24eb*}*i;9j2P#4Je|GfDRN1@I-cyRYAQOfMm{~pfgaX9L1o!Ub>`Le7Fza`CaH^%rP^<nRA@|PCp(1JP3tP=v6UXI&h#D)*WtMpNg*wQtQJ(#i2iLy#z2#Bikd7QLMzTDs>0Vsw{RShEqD?A{vNBV~%XxgCi}Sd>8c#NXurhz~`+6N3ID&DqDXhE27InzYETuBHN!1)7w+C&{gg#7E0kjfA}}2gCJA_v264)?V<F|W=uZ*HXcT!%N!I$5Pd2k#|kvX1`ojFvz2-d3dh8{0ZD>kV%{S+887&-yX#oT+?P1$gtbh54%dt7$nJ~j+0*$ljt>B#K-dU~Rel73a&`cR*%1If)4k%>wQPd*?9*_a`@y%$7}JXov&t|V4dhk2X_Gi)i#suXpfL&Nz4!7g-O4y*i#u-r;wyamYB^7yZ0<oxu<uv7etQtsOhB6H8j7-uJKr0a%r^Jig8+fJP|Z@9s~ngnn{wY?k7NdVS$)2jPQF;>&LgQxHcB=GVq~tH`7QRz`VOnZKD%GTS!rGB2mF#WDP0<z^e3G|l?Q#6hs5}hFUgw~J^IFnl8rav&(B`Ic+QksDBB_EPmW&NF;4;C$!+F<LzAINz>M!uhKlu$hoP0*MwE00wTdS>p`?jFT29`$@2>f7*iY|ZIpXK>-t?<z7YqIG+CjD}+Jybgu1KTr_4$7Ve?EWr-Wxx!?(jHCo6y<gt{4i_2l<>qn<immDVpi%1oJ?ic;*h;a5%+;R2Bm>R8V0tIz5L5X7O+!!4A%1EQ;Vw$i;EfQB1@|&+i-ejP>U~y?yrQm*>Iv|Kt4ozr6YDi`V`LBJtLEoZy}C)82;@A*`JYJAFCU*!cuDx6TZqcF2TZ6l?kNSdMW74WuZIi{pPyPR3u70mOSFSJ&(>ptXi?E2LB1pw1@r@8`v}=+3N4_o>|-N4eAl%k%I5qu34ZpE^`}UzJ656Pd$Pz;H_Fa2ZvJuDwvd!CVTl*2K8XoI7iTPmzZu3I2@dUf4|<^RcSm+kG7W`^6h9b^vD+XO#}WgU-I5qfcCnW3G>->kQiA3nT0`SMpXIH{E_+m%HntN^`Q{>PFqn;N0*l(@_6bbZvoCXS=To-c#{VmEEpwce`U+xR+O{$g?MHPw3n)zV_#0$!F)QD<HUc_g`{54_W)(eLNQA;QcZey$VbB${Dq9x1>Xd$FCf-7g*34L)@Zwe?`m5;nXaaU;6tJsnBsd5ACfs;tFEqp*fTFUO1--EZ=D3U1Sp~bl}g}O*HcMnb<kdfG8q~)w>W%7t%M)HaRA|NUp~qa5R2mhCVJFnI)9x^r+)M^G}_9;<MPDQ{T-4l^i^w-JX-@TqtqqB=mDjapxFx2SEQAbaiEdJ(KaM4lS1n9ygSltE1Y;R*YT1>Nb`GCnm|`O}X^wap&$CA{IZoJWHNT<6IOd&Su3s7w?0y&vfnJ#r)~%=~7$@D4l-lSX$Mok#lLO#!{kKYyWI;r+YSM#g30I<ye6{n$*Ss40+;vdAhM2v{CLF8YpqBvj<Ke%Y9(1Ol47i{7`#$xCCNDwHe|5am&L%1Vv5j10=OaNINvO-Oxn6qBG6###~IQqt$^@=icS?V>x_0vAT8Tq2q~la|&ax56*GJ2D5nS9ZaliyDT+wR<R9ci+*8vRfQ=diPD6Q0*hn3XE+j#IdapIY~qZhWn1%8y_ZcZX{)4ce*6g;LRIpPx$y5=vEK7v@5qYd_rzK};{!X$L@9s&n3KVLI6ayU79HOCst%`dZXo{SfrDp0&>f08oa!86)6Z@uE5HVUPk9|y+h#?>UhG39J1>sgfP2BM%vEc@7jT#H!QHj9vp;^p_T0m}dym+aE1wiOwF1Ii?nOTnU@v(`p>z7p3hyFjAiQ_rR53$#b=6?(Gv{;tkIE9l{VVEB<D=^|cvTy2Gdik<z1<cK39S_mn-OHWiG#Z>lc=OT3~{)Oqk*z6tdp(6VJ$xhLw7Jlh2u-&*TDZwN4g75;P7p32JN`F0DPJ7MEHl))V>QuYlC>fq@wP<I)Pm-eP-=f$KeW9L#a5Sv+JB{D)vMN9Lpiuq1aordWA(Nh2mapK?FBNcWZ?9b{o0vO<afxEm)M26nXg<?$Y+K2H;97ynLk5C$wIXfnw3Skg%7aD$b{~W&dD1rdA1qMFQAagE5V3Xz?AaoZ@sO<raT%@8Ac}V*SbO51~ELnILr4fhwb6E&I+o0t)tDfovgLj5%M?L4wxCQxDBmv1@A3(=@1Aath33fdd%(zz`A_#(H%%JH_M4_S^olz~Epg%KGXoA|RMybTSl=KSY>|`RsIQsWTKE_NaA0$g4YY^tHY2UC;nvN9K0$`uX=i(#16SHm^FGvwS;;2dv%i`0^SxQ85>w!5epOgX`==9^gp|hMD-P4-o8%uifm3N)wYKk@V#Gww)zGuTwpG49AC$4EgFLYbdFHjg+%y!`x2O(Xxsejf_Hn>p#N;4_mc@>jH+6T{mhxmG47WAB(}V%hlm{?QkW&8m@F*bZOS5Gqvv&I6ws+h*R<S3;XmPOpejk6`-d_0($xopxNoMD4mXi@+LQpk@IGx9x|c9fjP2Y^8Ud5>gwIuDc!R{GxzLt>3N^5hVuV91794^xXTf^M6F|iQsKjpXyYn@VIeZI_>rVt#>W7^A3MfIZ^lqpa9#L!=A86Iyt)*~5n_IR8@#*VfnJ)#Hahi)j-<nndU)x(da(C**dT#cx%)x%2SVfz@K0!W$kQFpf(-tKhqwRHPb|5hP8iv<l%$ccdI={0?df4?Pd^VDP4Pp>oqi6yS8v~rfb&JsX8^@jcpOh5tB!+t-{%fT5Sh2WPCoDf9Rvq;+=)<(AI{-E>BX8|6o*32!khs8YnzP+ogKjL1bhInDLe+6u231Ntb~1=l5nBmVHw8il9i9gUa}%a*0FEh2s2emJ_5CKnd(osh=BpW2lFdyS4E+42Te5m#T;h9(M_}CXgu_3yDX}F?yWKA<d^<cG4}I?4osu5ZX&~>Q^K<+>8yUT?L9#$DClz4awPaGI>$K40sj84ukh}n^XD*>L~k+(BNIL+QJu?(2ICIo9w3kmo4%vazNo6$njiuvPB|5xL(O=lU})mhvAIYD_iMNwMRk-Ic_>qSh+^|z_%w7RTZRSmwk|*9vb>CQ_e0-a4G#3;0H*8a7Ele%$zf}pfBVY0$4(RZ%R=gR+}eF;cB_nynhvQGT`_beF`D=M5$|yQ6%Fylo$lThyqbs`lW{`!j#+%{2QF(cv4Xk6rT$i9PTqOk*B-k?e58jza)QW-x2M7eK+mc5;TyelZ$wWkyr!qsS2e!Dqu-}fwJQmaq76P?=X8@*tL}63uZF$M@dyyQ7#7UV7M)L|fAG1a6QgfXFS@}kP5o<Q6`_tj0ob|Md%@m5`L5^bvv=^A%X{<!D=u3)@&@*SX(#r53pC^@UdP@%LBa6xdxUI1tn13`S6&S6UycXvJsDUpy#gaEr4_S2{hVwY{qY0c6vp0f=j0tvg7kKt{Zc)D`@e#>KfMm#JbU-<{P|oFHkdvA9pBN*2iDV6E6d=Uv&qxhQ+rPxMxOZ!F1a7g99sGj0HEM9EQ~212n&}*5LqxM0JA9+Nk$}rIOTNFKXYBP$mGP`I^_+W<O5%<OnOF&d!}0K(6PNKcI2A1l$3+}{_V3DuV1|WF@12|J7S>|loT<4#9+r4Age8dnsn}-TrU7eR1(y57Q+H>ouJDeQZ$spG{vPcP1S+zyiT6{Us-bPQ~"""
SCRIPT.write_bytes(zlib.decompress(base64.b85decode(PAYLOAD.encode("ascii"))))
print(f"Created {SCRIPT} ({SCRIPT.stat().st_size:,} bytes)", flush=True)

# Only training inputs are required here. Evaluation is deliberately separate.
required_files = [
    "training_triplets.csv",
    "synthetic_pairs_filtered.csv",
    "Signal-Android_Mistral_Ready.csv",
    "Anki-Android_Mistral_Ready.csv",
    "k-9_Mistral_Ready.csv",
    "brave_issues.csv",
]
input_root = Path("/kaggle/input")
all_input_files = [p for p in input_root.rglob("*.csv") if p.is_file()]

def canonical_stem(path):
    return re.sub(r"\s*\(\d+\)$", "", path.stem.lower())

missing = []
resolved = {}
for required in required_files:
    wanted = Path(required).stem.lower()
    matches = [p for p in all_input_files if canonical_stem(p) == wanted]
    if matches:
        resolved[required] = matches[0]
    else:
        missing.append(required)

if missing:
    print("CSV files currently attached:", flush=True)
    for path in all_input_files:
        print(" ", path, flush=True)
    raise FileNotFoundError(
        "Missing TRAINING inputs: " + ", ".join(missing) +
        "\nAttach the correct version of oqtayhasanov/spladetrainingv12."
    )
print("All required training files found:", flush=True)
for name, path in resolved.items():
    print(f"  {name} -> {path}", flush=True)

# Require two real GPUs before training.
gpu_lines = [line for line in subprocess.check_output(
    ["nvidia-smi", "-L"], text=True
).splitlines() if line.strip()]
print("\n".join(gpu_lines), flush=True)
if len(gpu_lines) < 2:
    raise RuntimeError(
        f"This notebook needs GPU T4 x2, but Kaggle exposed {len(gpu_lines)} GPU(s)."
    )

# Validate all bottom-15 counts and reconstructed text before training.
print("Running data validation...", flush=True)
subprocess.run(
    [sys.executable, str(SCRIPT), "--seed", str(SEEDS[0]),
     "--num-workers", "0", "--dry-run"],
    check=True,
)
print(f"Data validation passed. Launching seeds {SEEDS}.", flush=True)

jobs = []
for gpu_index, seed in enumerate(SEEDS):
    run_dir = Path(f"/kaggle/working/splade_bottom15_seed{seed}")
    resume_file = run_dir / "checkpoints" / "resume_state.pt"
    completed_file = run_dir / "TRAINING_COMPLETED"

    if completed_file.exists():
        print(f"Seed {seed} training is already complete; skipping it.", flush=True)
        continue
    if run_dir.exists() and any(run_dir.iterdir()) and not resume_file.exists():
        raise RuntimeError(
            f"{run_dir} contains partial files but no completed-epoch resume state."
        )

    log_path = Path(f"/kaggle/working/seed_{seed}.log")
    log_handle = log_path.open("a" if resume_file.exists() else "w", buffering=1)
    env = os.environ.copy()
    env["CUDA_DEVICE_ORDER"] = "PCI_BUS_ID"
    env["CUDA_VISIBLE_DEVICES"] = str(gpu_index)
    env["PYTHONUNBUFFERED"] = "1"
    env["TOKENIZERS_PARALLELISM"] = "false"

    command = [
        sys.executable, str(SCRIPT),
        "--seed", str(seed),
        "--num-workers", "1",
        "--skip-eval",
    ]
    if resume_file.exists():
        command.append("--resume")

    process = subprocess.Popen(
        command, env=env, stdout=log_handle, stderr=subprocess.STDOUT
    )
    jobs.append((seed, gpu_index, process, log_handle, log_path))
    print(
        f"STARTED seed={seed} GPU={gpu_index} PID={process.pid} "
        f"log={log_path}", flush=True
    )

if not jobs:
    print("All requested seeds were already complete.", flush=True)
else:
    while any(process.poll() is None for _, _, process, _, _ in jobs):
        time.sleep(60)
        status_parts = []
        for seed, gpu_index, process, _, log_path in jobs:
            try:
                lines = log_path.read_text(errors="replace").splitlines()
                latest = lines[-1] if lines else "starting..."
            except Exception as error:
                latest = f"log unavailable: {error}"
            state = "RUNNING" if process.poll() is None else f"EXIT={process.returncode}"
            status_parts.append(f"seed {seed} {state} | {latest}")
        print(time.strftime("%Y-%m-%d %H:%M:%S"), flush=True)
        print("\n".join(status_parts), flush=True)

    failures = []
    for seed, gpu_index, process, log_handle, log_path in jobs:
        return_code = process.wait()
        log_handle.close()
        print(f"FINISHED seed={seed} exit={return_code} log={log_path}", flush=True)
        if return_code != 0:
            failures.append((seed, return_code, str(log_path)))

    if failures:
        raise RuntimeError(f"One or more runs failed: {failures}")

print(f"SUCCESS: seeds {SEEDS} finished training.", flush=True)
print(
    "Checkpoints: "
    + ", ".join(f"/kaggle/working/splade_bottom15_seed{seed}" for seed in SEEDS),
    flush=True,
)
print("Evaluation will be run separately.", flush=True)
```
